````python
import shap
import torch
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from cnn_model import VoltageCNN, VoltageDNN, VoltageDataset

from scipy.cluster.hierarchy import dendrogram, linkage
from scipy.spatial.distance import pdist, squareform
````

````python
data_normal = pd.read_csv('./data/voltage_data.csv')
data_outage = pd.read_csv('./data/voltage_data_outage.csv')

voltage_data = data_normal.values
voltage_data_outage = data_outage.values

n_buses = voltage_data.shape[1]  # Number of voltage buses
time_steps = voltage_data.shape[0]  # Number of time steps
````

````python
def get_matrix_from_data(data_normal, data_outage):
    # # Combine the two datasets
    # combined_data = pd.concat([data_normal, data_outage])
    combined_data = data_normal
    # combined_data = data_outage

    # Compute the correlation matrix
    corr_matrix = combined_data.corr()

    # Convert the correlation matrix to a distance matrix
    distance_matrix = 1 - corr_matrix

    # Perform hierarchical clustering
    Z = linkage(squareform(distance_matrix), method='average')
    return Z, corr_matrix
````

````python
Z, corr_matrix = get_matrix_from_data(data_normal, data_outage)

# Plot the dendrogram
plt.figure(figsize=(10, 7))
dendrogram(Z, labels=corr_matrix.columns, leaf_rotation=45)
# dendrogram(Z, leaf_rotation=45)
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Feature')
plt.ylabel('Distance')
plt.show()
````

# 初始化模型

````python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = VoltageCNN(n_buses)
````

````python
_ = model(torch.tensor(np.random.rand(50, n_buses), dtype=torch.float32))
model.to(device)
````

````python
model.load_state_dict(torch.load('./voltage_cnn_model.pth', weights_only=True))
model.eval()
````

````python
def model_wrapper(x):
    x = torch.tensor(x, dtype=torch.float32).to(device)
    with torch.no_grad():
        probability = torch.sigmoid(model(x))  # Convert logits to probabilities
        prediction = (probability > 0.5).float()  # Convert to binary labels
    return prediction.cpu().numpy().squeeze()
````

# 划分训练和测试集

````python
dataset = VoltageDataset(voltage_data, voltage_data_outage)
train_size = int(0.8 * len(dataset))
test_size = len(dataset) - train_size
train_dataset, test_dataset = torch.utils.data.random_split(dataset, [train_size, test_size], generator=torch.Generator().manual_seed(20))
feature_names = [f'Bus {i+1}' for i in range(1, n_buses + 1)]
````

````python
label_1_samples = []
label_0_samples = []
for data, label in train_dataset:
    if label.item() == 1:  # Filter label 1 samples
        label_1_samples.append(data.numpy())  # Convert to NumPy for SHAP
    else: 
        label_0_samples.append(data.numpy())

# Convert to NumPy array for SHAP compatibility
label_0_samples = np.array(label_0_samples)
label_1_samples = np.array(label_1_samples)

print(label_0_samples.shape)
print(label_1_samples.shape)
````

````python
background_mean = torch.tensor(label_0_samples, dtype=torch.float32).mean(dim=0)
# background_mean = torch.tensor(np.zeros_like(background_mean)+0.1, dtype=torch.float32)  # Use zero mean as background

# Ensure at least (n_buses + 1) samples
n_buses = background_mean.shape[0]
num_background_samples = n_buses + 1  # Ensure SHAP has enough samples

# Expand background by adding small Gaussian noise to the mean values
background = background_mean.unsqueeze(0).repeat(num_background_samples, 1)  # Repeat mean row
background += torch.randn_like(background) * 0.01  # Add small noise for diversity

# Convert to NumPy for SHAP
background = background.cpu().numpy()

""" Check this line of code """
background = np.tile(background, (background.shape[1] + 1, 1))  # Expand background size

background.shape
````

# Initialize SHAP Explainer

````python
masker = shap.maskers.Partition(background)
masker.clustering = Z  # 强制使用你的 partition_tree
explainer = shap.PartitionExplainer(model_wrapper, masker, feature_names=feature_names)
````

````python
for test_sample in label_1_samples:
    shap_values = explainer(test_sample.reshape(1, n_buses))
    # shap_values.shape
    if 'total_shap_values_1' not in locals():
        total_shap_values_1 = shap_values
    else:
        total_shap_values_1.values += shap_values.values
    
print(total_shap_values_1)
print(total_shap_values_1.shape)
````

````python
shap.plots.waterfall(total_shap_values_1[0]/len(label_1_samples))
````

````python
import random
from sklearn.metrics import f1_score
from sklearn.metrics import accuracy_score

number_of_max_instances = 20  # Number of samples to select
number_of_attempts = 100  # Number of attempts to select samples
f1_scores = [0] * number_of_max_instances
accuracy_scores = [0] * number_of_max_instances

for number_of_instances in range(1, number_of_max_instances + 1):
    f1 = .0
    accuracy = .0
    for attempt in range(number_of_attempts):

        selected_samples = random.sample(list(label_1_samples), number_of_instances)

        total_shap_values_1 = None
        for test_sample in selected_samples:
            shap_values = explainer(test_sample.reshape(1, n_buses))
            if total_shap_values_1 is None:
                total_shap_values_1 = shap_values
            else:
                total_shap_values_1.values += shap_values.values

        # Normalize the total SHAP values by the number of selected samples
        total_shap_values_1.values /= number_of_instances

        # print(total_shap_values_1)
        # print(total_shap_values_1.shape)

        order_descending = np.argsort(total_shap_values_1[0].values)[::-1]
        top_2_indices = order_descending[:2]+1
        coded_array = np.zeros(len(order_descending))
        coded_array[top_2_indices] = 1

        '''Change bus_id to the bus you want to detect'''
        bus_id = np.array([2,3])
        bus_index = bus_id - 1
        target = np.zeros(len(order_descending))
        target[bus_index] = 1

        # Compute the F1 score
        f1 += f1_score(target, coded_array)
        accuracy += accuracy_score(target, coded_array)
    accuracy_scores[number_of_instances-1]=(accuracy / number_of_attempts)
    f1_scores[number_of_instances-1]=(f1 / number_of_attempts)
    
f1_scores, accuracy_scores
````
